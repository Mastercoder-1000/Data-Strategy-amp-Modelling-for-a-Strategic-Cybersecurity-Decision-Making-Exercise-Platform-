In a large country, there are four political parties, and each citizen supports exactly one of them. We want to estimate the national support percentages for these four parties using a random sample, due to limited resources. The goal is to find the minimum sample size $N$ that ensures, with at least 95% confidence, that the margin of error for each party's estimated support does not exceed 0.5 percentage points (in other words, guarantee that the margin of error for each of the four support percentages is not greater than 0.5 percentage points). For example, if the true support for a party is 23%, the estimate should fall between 22.5% and 23.5%.

**Scale:** 0.5 percentage points is 0.005 on the probability scale. If you work with proportions (0.23, not 23), the error limit is 0.005. If you work with percentages (23, not 0.23), the error limit is 0.5. Do not mix the two.

For this assignment, we will computationally estimate the required sample size under the assumption that the true support levels are known. Although this is unrealistic in practice, it allows us to study the required sample size under controlled conditions. We will consider two true distributions: **(I) 10%, 20%, 30%, and 40%**, and **(II) an even distribution of 25% for each party.**

**How many correlated pairs did we make? Justify the answer by constructing a permutation test. It is fine if your reported number of pairs is not exactly correct, the reasoning is more important than the answer.**

Here is how it works: shuffle the values within the columns, calculate all the absolute correlations between pairs of columns, and pick the highest one as our test statistic. Repeat this 100 times to create a range of test statistics (null distribution). The biggest value from these iterations becomes our significance threshold. Then, in the original data, find the absolute correlations; anything stronger than the threshold is treated as a potentially real correlation.
