**a. Read the dataset description. Who produced tha data? When, how, and for what purpose?**

**Answer:** 

CISA Cybersecurity Scenarios: Created by the U.S. Cybersecurity and Infrastructure Security Agency (CISA). The selected manuals were updated in 2023–2024 and contain fictional cybersecurity incidents and discussion questions to help organizations practice responding to cyber threats.

MITRE ATT&CK: Developed by MITRE and first publicly released in 2015. It is regularly updated using information about real-world cyberattacks and contains attack techniques and possible ways to prevent or reduce their impact.

Our main stakeholder is RiskSight OÜ, which develops cybersecurity training exercises for businesses. Their goal is to improve how exercises are organized and help users find suitable scenarios based on threats, difficulty, and required skills.

Our analysis could help RiskSight decide which exercises to recommend to different users. RiskSight and its customers would benefit from more relevant training. However, inaccurate recommendations could result in users practicing scenarios that do not match their training needs.

The second step of CRISP-DM is data understanding. The goal is to collect, describe, explore, and verify the quality of the data to ensure it’s suitable for the analytical or modeling tasks ahead.

**a. Describe the data. Number of rows and columns, what does one row represent.**

```python
import pandas as pd
from pathlib import Path

data_dir = Path("data/processed")

events = pd.read_csv(data_dir / "cisa_scenario_injects.csv")
classified = pd.read_csv(data_dir / "cisa_classified_events.csv")
questions = pd.read_csv(data_dir / "cisa_discussion_questions.csv")

print("Events:", events.shape)
print("Classified events:", classified.shape)
print("Questions:", questions.shape)
```

**Answer:** 

Scenario events: 41 rows and 6 columns. Each row represents one event in a cybersecurity scenario.

Classified events: 41 rows and 8 columns. This contains the same events, with two additional columns describing the threat type and business impact.

Discussion questions: 271 rows and 9 columns. Each row represents one discussion question from an exercise.

**b. Take a look at the data dictionaries or metadata to understand the meaning of each variable.**

Make sure to differenciate between technical type and meaning (integers for categorical, strings for dates, etc).

Find out the allowed values or range for each variable, check how categories are coded, and missing values represented. Think of the role of the variable in the project (target, candidate predictor, identifier, date/time, or not relevant).

If you can, try to find where the documentation and the data disagree.

You can do this as a text paragraph, a list, or as a table suggested below.


```python
for name, df in {
    "Scenario events": events,
    "Classified events": classified,
    "Discussion questions": questions
}.items():
    print(f"\n{name}")

    display(pd.DataFrame({
        "Data type": df.dtypes,
        "Unique values": df.nunique(),
        "Example value": df.iloc[0]
    }))
```

**Answer:** 

Most variables are text or categorical. The datasets contain the source manual, section, scenario day, event descriptions, and discussion questions. The inject_id and question_id columns are unique identifiers, while source_paragraph is a numerical reference to the original Word document.

The classified events dataset also contains threat_type (24 unique values) and business_impact (41 unique descriptions), which we assigned ourselves based on the CISA manuals.

The discussion questions dataset has three additional columns (domain_category, difficulty_level, and required_skill), which are currently empty because we have not classified the questions yet.

**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import matplotlib.pyplot as plt

# Scenario events by manual
events["manual_title"].value_counts().plot(kind="bar")

plt.title("Scenario events by CISA manual")
plt.xlabel("Manual")
plt.ylabel("Number of events")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()

# Discussion questions by manual
questions["manual_title"].value_counts().plot(kind="bar")

plt.title("Discussion questions by CISA manual")
plt.xlabel("Manual")
plt.ylabel("Number of questions")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

**Answer:** 

The graphs show that the number of extracted events and discussion questions varies across the five CISA manuals.

The Insider Threat manual has the most scenario events (12), while the Executives manual has the fewest (2). Healthcare has the most discussion questions (61), while Vendor and Executives have the fewest (49 each).

Overall, the manuals differ in structure and content. The number of scenario events does not necessarily correspond to the number of discussion questions. For example, Executives has only 2 events but 49 questions.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
for name, df in {
    "Scenario events": events,
    "Classified events": classified,
    "Discussion questions": questions
}.items():
    print(f"\n{name}")

    print("Missing values:")
    print(df.isna().sum())

    print("Duplicate IDs:", df[
        "question_id" if name == "Discussion questions" else "inject_id"
    ].duplicated().sum())
```

**Answer:** 

The datasets contain no duplicate IDs or missing event descriptions or question texts. The missing `day` values (2 for events and 49 for questions) are expected because the Executives manual does not use days.

All 41 events have been classified, but the three discussion question classification columns are still empty. Some question texts are repeated across exercises, which is not necessarily an error.

The main possible data quality issues are errors in automatic extraction and inconsistencies in manually assigned labels.

However, the CISA scenarios are fictional, and we do not have participant performance data or verified difficulty levels. This makes it harder to evaluate exercise effectiveness or develop accurate recommendations.
